```python
# 모델링 및 전처리 라이브러리
from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import f1_score
from sklearn.ensemble import RandomForestClassifier
from lightgbm import LGBMClassifier
from lightgbm import early_stopping
```

```python
# # 'gender' 컬럼 라벨 인코딩
# le = LabelEncoder()
# train['gender'] = le.fit_transform(train['gender'])
# test['gender'] = le.transform(test['gender'])

# # 'subscription_type' 컬럼 라벨 인코딩
# train['subscription_type'] = le.fit_transform(train['subscription_type'])
# test['subscription_type'] = le.transform(test['subscription_type'])

# # 'age_group' 컬럼 라벨 인코딩
# train['age_group'] = le.fit_transform(train['age_group'])
# test['age_group'] = le.transform(test['age_group'])

# 결측치 처리 (만약 결측치가 있다면)
# 현재 데이터셋에는 결측치가 없다고 가정하고 코드를 진행합니다.
# 만약 있다면 df.isnull().sum() 으로 확인 후, 적절한 방법으로 처리해야 합니다.

# ID와 타겟 변수 분리
train_X = train.drop(['ID', 'support_needs'], axis=1)
train_y = train['support_needs']
test_X = test.drop('ID', axis=1)

# 범주형 변수 리스트 (새로 만든 'age_group' 포함)
categorical_features = ['gender', 'subscription_type'] # 해도 정확도 오히려 떨어져서 age_group 뺌
for feature in categorical_features:
    le = LabelEncoder()
    # train과 test 데이터를 합쳐서 모든 범주를 학습시킵니다.
    combined_data = pd.concat([train_X[feature], test_X[feature]]).astype(str)
    le.fit(combined_data)
    train_X[feature] = le.transform(train_X[feature].astype(str))
    test_X[feature] = le.transform(test_X[feature].astype(str))
```

####RandomForest Model 사용 (but. lgbm보다 안 좋음)

```python
# 4. 모델 학습 및 예측 RandomForest
print("RandomForest 모델 학습을 시작합니다...")
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
oof_preds = np.zeros(train_X.shape[0])
test_preds = np.zeros((test_X.shape[0], 3))

model = RandomForestClassifier(n_estimators=200,
                               random_state=42,
                               class_weight='balanced',
                               n_jobs=-1) # CPU 코어를 모두 사용하여 학습 속도 향상

for fold, (train_idx, val_idx) in enumerate(skf.split(train_X, train_y)):
    print(f"--- Fold {fold+1} ---")
    X_train, y_train = train_X.iloc[train_idx], train_y.iloc[train_idx]
    X_val, y_val = train_X.iloc[val_idx], train_y.iloc[val_idx]

    model.fit(X_train, y_train)

    oof_preds[val_idx] = model.predict(X_val)
    test_preds += model.predict_proba(test_X) / skf.n_splits

# 최종 OOF Macro F1-Score 확인
oof_f1_macro = f1_score(train_y, oof_preds, average='macro')
print(f"\nOOF Macro F1-Score with Age Binning (RandomForest): {oof_f1_macro:.4f}")

# 테스트 데이터 최종 예측 (확률이 가장 높은 클래스를 선택)
# test_preds는 각 클래스에 대한 확률이므로, 가장 높은 확률의 인덱스를 선택하여 최종 예측값으로 사용합니다.
final_predictions = np.argmax(test_preds, axis=1)

sample_submission = pd.read_csv('sample_submission(dacon).csv')

# 제출 파일 생성
sample_submission['support_needs'] = final_predictions
sample_submission.to_csv('submission_RandomForest.csv', index=False)

print("\nSubmission file 'submission_RandomForest.csv' has been created successfully.")
```

####Light GBM 사용

LGBMClassifier()에 num_leaves, max_depth 추가

```python
# Stratified K-Fold 교차 검증 설정
# StratifiedKFold는 클래스 비율을 유지하면서 데이터를 분할하여 불균형한 데이터셋에 적합합니다.
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
oof_preds = np.zeros(train_X.shape[0])
test_preds = np.zeros((test_X.shape[0], 3)) # (샘플 수 , 클래스 수) 형태로 초기화

# LightGBM 모델 설정
# objective='multiclass'로 다중 클래스 분류를 지정하고, num_class로 클래스 수를 지정합니다.
lgbm = LGBMClassifier(objective='multiclass',
                      num_leaves=31,
                      max_depth=-1,
                      num_class=3,
                      random_state=42,
                      n_estimators=1000, # early stopping 으로 인해 굳이 여러번 실험할 필요x
                      learning_rate=0.05,
                      # 클래스 불균형을 고려한 가중치 설정
                      class_weight='balanced')

# 교차 검증 진행
for fold, (train_idx, val_idx) in enumerate(skf.split(train_X, train_y)):
    print(f"--- Fold {fold+1} ---")

    # 학습/검증 데이터 분리
    X_train, y_train = train_X.iloc[train_idx], train_y.iloc[train_idx]
    X_val, y_val = train_X.iloc[val_idx], train_y.iloc[val_idx]

    # 모델 학습 (Early Stopping 적용)
    # val_y에 대한 평가 지표를 'multi_logloss'로 설정하여 early stopping을 진행합니다.
    lgbm.fit(X_train, y_train,
             eval_set=[(X_val, y_val)],
             eval_metric='multi_logloss',
             callbacks=[early_stopping(100, verbose=False)])

    # OOF(Out-of-Fold) 예측
    oof_preds[val_idx] = lgbm.predict(X_val)

    # 테스트 데이터 예측
    test_preds += lgbm.predict_proba(test_X) / skf.n_splits

# 최종 OOF Macro F1-Score 확인
oof_f1_macro = f1_score(train_y, oof_preds, average='macro')
print(f"\nOOF Macro F1-Score: {oof_f1_macro:.4f}")


# 테스트 데이터 최종 예측 (확률이 가장 높은 클래스를 선택)
# test_preds는 각 클래스에 대한 확률이므로, 가장 높은 확률의 인덱스를 선택하여 최종 예측값으로 사용합니다.
final_predictions = np.argmax(test_preds, axis=1)

sample_submission = pd.read_csv('sample_submission(dacon).csv')

# 제출 파일 생성
sample_submission['support_needs'] = final_predictions
sample_submission.to_csv('submission_LGBMversion0.1.csv', index=False)

print("\nSubmission file 'submission_LGBMversion0.1.csv' has been created successfully.")
```

####하이퍼파라미터 50번 돌리고 최적의 하이퍼파라미터 찾는 함수 Optuna

```python
# 하이퍼파라미터 50번 돌리고 최적의 하이퍼파라미터 찾는 함수
# --- 1. Objective 함수 정의 ---
def objective(trial):
    # 하이퍼파라미터 탐색 공간 정의
    params = {
        'objective': 'multiclass',
        'num_class': 3,
        'metric': 'multi_logloss',
        'random_state': 42,
        'n_estimators': 1000,
        'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.1),
        'num_leaves': trial.suggest_int('num_leaves', 20, 300),
        'max_depth': trial.suggest_int('max_depth', 3, 12),
        'min_child_samples': trial.suggest_int('min_child_samples', 5, 100),
        'subsample': trial.suggest_float('subsample', 0.6, 1.0),
        'colsample_bytree': trial.suggest_float('colsample_bytree', 0.6, 1.0),
        'reg_alpha': trial.suggest_float('reg_alpha', 0.0, 1.0),
        'reg_lambda': trial.suggest_float('reg_lambda', 0.0, 1.0),
        'class_weight': 'balanced'
    }

    # K-Fold 교차 검증으로 성능 평가
    skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
    f1_scores = []
    for train_idx, val_idx in skf.split(train_X, train_y):
        X_train, y_train = train_X.iloc[train_idx], train_y.iloc[train_idx]
        X_val, y_val = train_X.iloc[val_idx], train_y.iloc[val_idx]

        model = LGBMClassifier(**params)
        model.fit(X_train, y_train,
                  eval_set=[(X_val, y_val)],
                  eval_metric='multi_logloss',
                  callbacks=[]) # early_stopping은 Optuna가 n_estimators를 다루도록 잠시 비활성화

        preds = model.predict(X_val)
        f1 = f1_score(y_val, preds, average='macro')
        f1_scores.append(f1)

    # 교차 검증 F1 점수의 평균을 반환 (Optuna는 이 값을 최대화하는 방향으로 탐색)
    return np.mean(f1_scores)

# --- 2. 최적화 실행 ---
# 'maximize' 방향으로 Macro F1 Score를 최적화
study = optuna.create_study(direction='maximize')
# 50번의 시도를 통해 최적의 파라미터를 탐색
study.optimize(objective, n_trials=50)

# --- 3. 결과 확인 ---
print("Number of finished trials: ", len(study.trials))
print("Best trial:")
trial = study.best_trial

print("  Value: {:.4f}".format(trial.value))
print("  Params: ")
for key, value in trial.params.items():
    print("    {}: {}".format(key, value))

# 찾은 최적의 파라미터로 모델을 다시 학습하여 최종 예측 파일을 생성할 수 있습니다.
best_params = trial.params
```

##### XGBoost 최적 하이퍼파라미터 찾기

```python
import optuna
import pandas as pd
import numpy as np
from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import f1_score
from xgboost import XGBClassifier
import warnings

warnings.filterwarnings('ignore')

# --- 1. 데이터 로딩 및 최종 피처 엔지니어링 ---
# (이전 코드와 동일하게 데이터를 준비합니다)
train_df = pd.read_csv('train(dacon).csv')
test_df = pd.read_csv('test(dacon).csv')

for df in [train_df, test_df]:
    df['after_interaction_div_tenure'] = df['after_interaction'] / (df['tenure'] + 1)
    df['age_x_tenure'] = df['age'] * df['tenure']

train_X = train_df.drop(['ID', 'support_needs'], axis=1)
train_y = train_df['support_needs']
# 이하 전처리 코드는 동일...
categorical_features = ['gender', 'subscription_type']
for feature in categorical_features:
    le = LabelEncoder()
    combined_data = pd.concat([train_X[feature], test_df.drop('ID', axis=1)[feature]]).astype(str)
    le.fit(combined_data)
    train_X[feature] = le.transform(train_X[feature].astype(str))
    test_X[feature] = test_df.drop('ID', axis=1)[feature].astype(str)
    test_X[feature] = le.transform(test_X[feature])
# ----------------------------------------------------


# --- 2. XGBoost Objective 함수 정의 (수정) ---
def objective_xgb(trial):
    params = {
        # 🚨 n_estimators를 Optuna가 찾도록 탐색 공간에 추가
        'n_estimators': trial.suggest_int('n_estimators', 200, 2000, step=100),
        'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.1),
        'max_depth': trial.suggest_int('max_depth', 3, 12),
        'subsample': trial.suggest_float('subsample', 0.6, 1.0),
        'colsample_bytree': trial.suggest_float('colsample_bytree', 0.6, 1.0),
        'gamma': trial.suggest_float('gamma', 0, 5),
        'min_child_weight': trial.suggest_int('min_child_weight', 1, 10),
        'objective': 'multi:softprob',
        'num_class': 3,
        'eval_metric': 'mlogloss',
        'random_state': 42,
        'tree_method': 'hist',
        'device': 'cpu'
    }

    skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
    f1_scores = []
    for train_idx, val_idx in skf.split(train_X, train_y):
        X_train, y_train = train_X.iloc[train_idx], train_y.iloc[train_idx]
        X_val, y_val = train_X.iloc[val_idx], train_y.iloc[val_idx]

        model = XGBClassifier(**params)
        # 🚨 early_stopping_rounds 없이 학습
        model.fit(X_train, y_train)

        preds = model.predict(X_val)
        f1 = f1_score(y_val, preds, average='macro')
        f1_scores.append(f1)

    return np.mean(f1_scores)

# --- 3. 최적화 실행 ---
study_xgb = optuna.create_study(direction='maximize')
study_xgb.optimize(objective_xgb, n_trials=50)

# --- 4. 결과 확인 ---
print("Best trial for XGBoost:")
trial = study_xgb.best_trial
print("  Value: {:.4f}".format(trial.value))
print("  Params: ")
for key, value in trial.params.items():
    print("    {}: {}".format(key, value))
```

```python
import optuna
import pandas as pd
import numpy as np
from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import f1_score
from catboost import CatBoostClassifier
import warnings

warnings.filterwarnings('ignore')

# --- 1. 데이터 로딩 및 최종 피처 엔지니어링 ---
# (이전 코드와 동일하게 데이터를 준비합니다)
train_df = pd.read_csv('train(dacon).csv')
test_df = pd.read_csv('test(dacon).csv')

for df in [train_df, test_df]:
    df['after_interaction_div_tenure'] = df['after_interaction'] / (df['tenure'] + 1)
    df['age_x_tenure'] = df['age'] * df['tenure']

train_X = train_df.drop(['ID', 'support_needs'], axis=1)
train_y = train_df['support_needs']
test_X = test_df.drop('ID', axis=1)

categorical_features = ['gender', 'subscription_type']
for feature in categorical_features:
    le = LabelEncoder()
    combined_data = pd.concat([train_X[feature], test_df[feature]]).astype(str)
    le.fit(combined_data)
    train_X[feature] = le.transform(train_X[feature].astype(str))
    test_X[feature] = le.transform(test_df[feature].astype(str))
# ----------------------------------------------------


# --- 2. CatBoost Objective 함수 정의 ---
def objective_catboost(trial):
    # CatBoost 하이퍼파라미터 탐색 공간 정의
    params = {
        'objective': 'MultiClass',
        'iterations': trial.suggest_int('iterations', 500, 2000),
        'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.1),
        'depth': trial.suggest_int('depth', 4, 10),
        'l2_leaf_reg': trial.suggest_float('l2_leaf_reg', 1, 10), # L2 정규화
        # 'subsample': trial.suggest_float('subsample', 0.6, 1.0), # LightGBM의 subsample과 유사
        'random_strength': trial.suggest_float('random_strength', 1e-9, 10, log=True), # 무작위성 조절
        'bagging_temperature': trial.suggest_float('bagging_temperature', 0, 1), # 배깅 강도 조절
        'border_count': trial.suggest_int('border_count', 32, 255), # 수치형 피처를 나누는 구간 수
        'verbose': 0,
        'random_seed': 42
    }

    skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
    f1_scores = []
    for train_idx, val_idx in skf.split(train_X, train_y):
        X_train, y_train = train_X.iloc[train_idx], train_y.iloc[train_idx]
        X_val, y_val = train_X.iloc[val_idx], train_y.iloc[val_idx]

        model = CatBoostClassifier(**params)
        model.fit(X_train, y_train,
                  eval_set=[(X_val, y_val)],
                  early_stopping_rounds=100,
                  verbose=0)

        preds = model.predict(X_val)
        f1 = f1_score(y_val, preds, average='macro')
        f1_scores.append(f1)

    return np.mean(f1_scores)

# --- 3. 최적화 실행 ---
study_catboost = optuna.create_study(direction='maximize')
study_catboost.optimize(objective_catboost, n_trials=50) # 시간 여유에 따라 n_trials 조절

# --- 4. 결과 확인 ---
print("\nBest trial for CatBoost:")
trial = study_catboost.best_trial
print("  Value: {:.4f}".format(trial.value))
print("  Params: ")
for key, value in trial.params.items():
    print("    {}: {}".format(key, value))
```
